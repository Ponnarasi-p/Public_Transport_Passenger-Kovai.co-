```python
Technical Report on ARIMA Model for Public Transport Passenger Forecasting
1. Introduction
This report presents the use of the ARIMA (AutoRegressive Integrated Moving Average) model to forecast passenger counts across various transport service types such as Local Route, Light Rail, Peak Service, Rapid Route, and School for the next 7 days. The primary objective is to identify trends, seasonality, and make accurate short-term predictions that can assist in operational planning and scheduling.
2.Algorithm Overview
The ARIMA model is one of the most popular time series forecasting methods. It combines three key components:
•	AR (AutoRegressive): Uses the dependency between an observation and a number of lagged observations.
•	I (Integrated): Uses differencing of raw observations to make the time series stationary (removes trend or seasonality).
•	MA (Moving Average): Uses the dependency between an observation and a residual error from a moving average model applied to lagged observations.
The general form of the model is represented as ARIMA(p, d, q) where:
•	p = number of lag observations in the model (AR term)
•	d = number of times that the raw observations are differenced (to remove trend)
•	q = size of the moving average window (MA term)
3. Model Selection and Parameter Tuning
To select the optimal parameters (p, d, q), Akaike Information Criterion (AIC) was used — a lower AIC indicates a better-fit model.
After evaluating several combinations, ARIMA(2, 0, 2) achieved the lowest AIC value (≈ 41935), suggesting the best balance between model accuracy and complexity.
•	p = 2: The model depends on the previous two lag values.
•	d = 0: The data was already stationary (confirmed using the Augmented Dickey-Fuller test).
•	q = 2: The moving average term uses two previous forecast errors.
 4. Model Evaluation
•	The model’s forecasting performance was assessed using:
•	Mean Absolute Error (MAE): 17023.36
•	Root Mean Squared Error (RMSE): 18355.08
•	These metrics indicate moderate prediction accuracy, which is acceptable for high-variance real-world transport data.
5. Forecast Results
The ARIMA(2, 0, 2) model successfully forecasted passenger demand for the next 7 days.
Forecasts show:
•	Stable demand for Local Route and Rapid Route services.
•	Slight variations for Light Rail and Peak Service.
•	Decline in School Route usage (possibly due to holidays).
•	6. Conclusion
The ARIMA model effectively captured short-term patterns and provided reliable forecasts for operational planning. Although models like Prophet may handle seasonality more robustly, ARIMA remains a strong choice for datasets with stable, stationary characteristics.
Further accuracy improvements could be achieved by incorporating exogenous variables (e.g., temperature, holidays, or events) using SARIMAX or Prophet.




```
